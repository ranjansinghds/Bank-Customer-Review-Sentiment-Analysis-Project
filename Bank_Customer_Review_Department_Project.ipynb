# TASK #3: Data Cleaning

```python
# dropping not useful column
banks_df = banks_df.drop(['author', 'date', 'location', 'bank', 'star', 'like', 'review_length'], axis=1)
```

```python
banks_df.head()
```

```python
import re
import string
```

```python
# Text preprocessing (preparing the review text for NLP.)
def clean_text(text):

    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
banks_df['clean_text'] = banks_df['text'].apply(clean_text)
```

```python
banks_df[['text', 'clean_text']]
```

# TASK #4: Spliting a data for TF-IDF Vectorization

```python
X = banks_df['clean_text']
y = banks_df['sentiment']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2, stratify=y)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

```python
# Using TF-IDF Vectorization to convert text into nemerical features.

tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=20000,
    ngram_range = (1, 2),
    min_df= 2,
    max_df = 0.95
)
```

```python
tfidf
```

```python
# fitting only on training data
X_train_tfidf = tfidf.fit_transform(X_train)
```

```python
# doing transform on test data
X_test_tfidf = tfidf.transform(X_test)
```

 **Why fit only on training data?**<br>
Because fitting the vectorizer on the complete dataset can cause data leakage.

# TASK #5: TRAIN A NAIVE BAYES CLASSIFIER MODEL

```python
X_train_tfidf.shape
```

```python
X_test_tfidf.shape
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train_tfidf, y_train)
```

```python
nb_pred = nb_model.predict(X_test_tfidf)
```

```python
nb_pred
```

# Evaluate Naive Bayes Trained Model

```python
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, roc_auc_score)
```

```python
nb_accuracy = accuracy_score(y_test, nb_pred)
print('Naive Bayes Accuracy:', nb_accuracy)
```

```python
cm = confusion_matrix(y_test, nb_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Negative', 'Positive'])
disp.plot()

plt.title('Naive BAyes Coonfusion Matrix')
plt.show()
```

```python
print(classification_report(y_test, nb_pred, target_names=['Negative', 'Positive']))
```
