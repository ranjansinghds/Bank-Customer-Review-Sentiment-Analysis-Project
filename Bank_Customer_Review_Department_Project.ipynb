```python
# Loading the dataset
banks_df = pd.read_csv('Banks.csv')
```

```python
# checking the missing value
plt.figure(figsize=(12,6))
sns.heatmap(banks_df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
# checking missing values
banks_df.isnull().sum()
```

```python
# checking unique values
banks_df.nunique()
```

```python
# Checking the text column
banks_df['text']
```

```python
# we need to drop that missing values
banks_df = banks_df.dropna(subset=['text'])
```

```python
# Droping empty reviews(string)
banks_df['text'] = banks_df['text'].astype(str).str.strip()
banks_df = banks_df[banks_df['text'] != ""]
```

```python
# checking star column
banks_df['star'].value_counts().sort_index()
```

```python
plt.figure(figsize=(10,4))
sns.countplot(x='star', data=banks_df, hue='star', palette='tab10' )
plt.title('Distribution of star Rating')
plt.xlabel("Star Rating")
plt.ylabel('Number of Reviews')
```

```python
# checking the number of bank s
banks_df['bank'].nunique()
```

```python
banks_df['bank'].value_counts()
```

```python
# Visualizing the top 15 banks
plt.figure(figsize=(15, 8))
banks_df['bank'].value_counts().head(15).plot(kind='bar')
plt.title('Top 15 Banks by Number of Reviews')
plt.xlabel('Bank')
plt.ylabel('Number of Reviews')

plt.xticks(rotation=90)
plt.show()
```
